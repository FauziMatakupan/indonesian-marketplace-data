# OLX Indonesia Computer & Laptop Listings

1,000 public listings from the OLX Indonesia category **Komputer & Laptop**, captured on 15 November 2026.

Files:
- `olx-1000.csv`: 1,000 rows, 19 columns, UTF-8
- `olx-1000.parquet`: same data, typed, smaller
- `data-dictionary.csv`: column names, types, missing counts, examples
- `olx-listing.schema.json`: JSON Schema for one listing

Source page: https://matakupanlabs.com/dataset/olx-indonesia/

```python
import pandas as pd

pd.set_option('display.max_columns', None)
df = pd.read_csv('/kaggle/input/olx-indonesia-computer-laptop-1000/olx-1000.csv')
df.shape
```

```python
df.head(3)
```

```python
summary = pd.DataFrame({
    'dtype': df.dtypes.astype(str),
    'non_null': df.notna().sum(),
    'missing': df.isna().sum(),
    'unique': df.nunique(dropna=True),
})
summary
```

## Price distribution

Prices are asking prices in IDR, not transaction prices.

```python
df['price'].describe().astype(int).to_frame().T
```

```python
(
    df.groupby('condition')['price']
      .agg(listings='count', median='median', mean='mean', min='min', max='max')
      .sort_values('median', ascending=False)
      .astype(int)
)
```
